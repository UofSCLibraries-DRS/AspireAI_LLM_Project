# Building NTLK corpus for English Words with Pronouns

```python
import nltk
import pandas as pd
from nltk.corpus import words
from nltk.tokenize import word_tokenize

# # Download NLTK data (only need to run once)
# nltk.download('words')
# nltk.download('punkt_tab')
```

```python
dataset = "simkins"
df = pd.read_csv(r'../data/mccray/original_data/McCray metadata.csv')
df = pd.read_csv(fr'../data/{dataset}/original_data/Simkins metadata.csv')


# Get the original NLTK words corpus
original_words = set(words.words())

# Tokenize and collect all words from Title + Description
extra_words = set()

for _, row in df.iterrows():
    text = f"{row['Title']} {row['Description']}"
    tokens = word_tokenize(text.lower())
    # Keep alphabetic tokens only
    extra_words.update([t for t in tokens if t.isalpha()])

# Merge both (base corpus + mccray corpus v2 addition)
combined_words = original_words.union(extra_words)

# Remove single letters except 'a' and 'i'
combined_words = {w for w in combined_words if len(w) > 1 or w in ('a', 'i')}
extra_words = {w for w in extra_words if len(w) > 1 or w in ('a', 'i')}

print(f"Original corpus size: {len(original_words):,}")
print(f"Extra words from DF: {len(extra_words):,}")
print(f"Combined total: {len(combined_words):,}")

# Example usage:
# Can now check if a word is in your combined vocabulary
print("\nVocab Check:")
print("she", "she" in combined_words)   # True
print("collaboration", "collaboration" in combined_words)  # True
print("mccray", "mccray" in combined_words)
print("carolina", "carolina" in combined_words)
print("john", "john" in combined_words)
print("adams", "adams" in combined_words)
print("george", "george" in combined_words)
print("i", "i" in combined_words)
print("a", "a" in combined_words)
print("z", "z" in combined_words)
print("j", "j" in combined_words)

with open("../data/mccray/combined_words_corpus.txt", "w") as f:
    for w in sorted(combined_words):
        f.write(w + "\n")

with open("../data/mccray/mccray_only_words_corpus.txt", "w") as f:
    for w in sorted(extra_words):
        f.write(w + "\n")     
```
