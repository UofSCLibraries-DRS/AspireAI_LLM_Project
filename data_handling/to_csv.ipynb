# To CSV

```python
import pandas as pd
import csv
```

```python
# input/output
input = r'../data/simkins/original_data/Simkins metadata.xlsx'
output = r'../data/simkins/original_data/Simkins metadata.csv'

def save_csv(df, title):
    df.to_csv(title,
          index=False,
          encoding='utf-8',
          quoting=csv.QUOTE_NONNUMERIC,   
          escapechar='\\',
          lineterminator='\n')
```

```python
df = pd.read_excel(input)

save_csv(df, output)
```
