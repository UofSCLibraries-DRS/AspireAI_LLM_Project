```python
import pandas as pd
import utils.csver as sv
from utils.ocr_cleaning import OCR_Clean as cn
```

```python
# Get original metadata
df = pd.read_csv(r"../data/mccray/original_data/McCray metadata.csv")
print(list(df))
```

```python
# Key words grab from subject column (remove repeats, output as list)
subjects = df.drop_duplicates(subset=["Subject"], keep='first')
descriptions = df.drop_duplicates(subset=["Description"], keep='first')

'''
replace '--' with ' '
replace ";" with "; "
'''
subjects = (
    subjects['Subject']
    .fillna("")
    .str.replace('--', ' ', regex=False)
    .str.replace(';;', ';', regex=False)
    .str.replace(';', '; ', regex=False)
    .str.replace("?", "", regex=False)
    .str.replace(r'[^a-zA-Z0-9\s.,!;:\'"()\-_/]+', '', regex=True)
)
subjects
```

```python
# output to look over
sv.save_csv(subjects, "subjects.csv")
```
