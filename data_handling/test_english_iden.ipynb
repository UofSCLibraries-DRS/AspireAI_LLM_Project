```python
import pandas as pd
import nltk
from nltk.corpus import words
import re
```

```python
# df
df = pd.read_csv("input/mini_test.csv")
sc_col = "Semi-clean Transcript"
nltk.download('words')
english_vocab = set(w.lower() for w in words.words())
```

```python
def percent_real_english_nltk(text):
    tokens = re.findall(r"[A-Za-z]+", text.lower())
    if not tokens:
        return 0
    real = [t for t in tokens if t in english_vocab]
    return 100 * len(real) / len(tokens)

# Apply to semi-clean column
print("Detecting language percentage for sc col...")
df['% English'] = df[sc_col].apply(percent_real_english_nltk)
df
```
