```python
import re, pandas as pd
from utils.ocr_cleaning import OCR_Clean

def clean_single_v2(text, quality=True, short_no_transcript=False):
    """Return (cleaned_text, sc_v2_len) for a single transcript string using the v2 logic."""
    # mimic DataFrame .where behavior: if quality True keep original
    if pd.notna(text) and quality:
        result = text
    else:
        if short_no_transcript:
            result = ""
        else:
            result = "" if text is None else str(text)
            # same sequence as notebook v2
            result = re.sub(OCR_Clean.non_ascii, "", result)
            result = re.sub(OCR_Clean.special_pattern_expanded, "", result)
            result = re.sub(OCR_Clean.letter_pattern_4, r"\\1", result)
            compiled = OCR_Clean.compile()
            result = OCR_Clean.clean_ocr_text(result, compiled)
    # normalize spaces (auto-newline)
    result = re.sub(r" {2,}", lambda m: "\n" if len(m.group(0)) > 3 else " ", result)
    sc_v2_len = len(result.replace(" ", ""))
    return result, sc_v2_len

# Example usage:
sample = ". , /7 / 77 /    LITHO U.S.A.\nMade in U. S. A.\n804    LITHO U.S.A."
cleaned, length = clean_single_v2(sample, quality=False, short_no_transcript=False)
print(cleaned)
print("sc_v2_len =", length)
```
