```python
import pandas as pd
import re, nltk, csv
from nltk.corpus import words, PlaintextCorpusReader
```

```python
# Get original metadata
# mccray = pd.read_csv(r"../../data/mccray/original_data/McCray metadata.csv")
# print(list(mccray))

def save_csv(df, title):
    df.to_csv(title,
          index=False,
          encoding='utf-8',
          quoting=csv.QUOTE_NONNUMERIC,   
          escapechar='\\',
          lineterminator='\n')

# # nltk download
# nltk.download('words')
english_vocab_base = set(w.lower() for w in words.words())


wordlist = PlaintextCorpusReader('../../data/mccray', 'combined_words_corpus.txt')
english_vocab_improved = set(w.lower() for w in wordlist.words())
```

```python
def percent_real_english_nltk(text, vocab):
    tokens = re.findall(r"[A-Za-z]+", text.lower())
    print("Tokens", tokens)
    if not tokens:
        return 0
    real = [t for t in tokens if t in vocab]
    return 100 * len(real) / len(tokens)

def real_english_nltk_dict(text, vocab):
    """Return dict showing which tokens are recognized or not (NLTK-style regex)."""
    if not isinstance(text, str) or not text.strip():
        return {"words": [], "not_words": []}

    tokens = re.findall(r"[A-Za-z]+", text.lower())
    word_count = len(tokens)
    words, not_words = [], []
    for t in tokens:
        (words if t in vocab else not_words).append(t)
    valid_word_count = len(words)
    return {"words": words, "not_words": not_words, "word_count":word_count, "valid_word_count":valid_word_count}

txt_1_cd_10 = 'If you have a boy interested in selling the LIGHTHOUSE (newspaper),  have him sign name with address below. The representative will make the  personal contact with the parent at his hone.  J. W. Witherspoon, Jr.  Principal  Reese, Calvin       911 Tree St.'

print(txt_1_cd_10)

print(percent_real_english_nltk(txt_1_cd_10, english_vocab_base))
print(real_english_nltk_dict(txt_1_cd_10, english_vocab_base), "\n")

print(percent_real_english_nltk(txt_1_cd_10, english_vocab_improved))
print(real_english_nltk_dict(txt_1_cd_10, english_vocab_improved))
```
